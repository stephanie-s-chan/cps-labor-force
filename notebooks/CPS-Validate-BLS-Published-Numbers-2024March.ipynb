[B] BLS CPS data methodology
1. Weight choice matters, and depends on the population. For example, `COMPWT` reproduces the national estimates; `WTVET`, which is controlled to Veterans Affairs veteran totals, reproduces the veteran and nonveteran figures.
2. Seasonal adjustment: Microdata match BLS's unadjusted series. Seasonally adjusted figures are not a valid benchmark for a single month of microdata.
3. Rates alone can mislead and levels are the stronger test.The veteran unemployment rate under `WTFINL` matched BLS (3.0%) even though every level was 7–11% too low, because weight errors largely cancel in a ratio.
4. Population totals are a control check. Matching the population confirms the universe and weights before any labor force statistic is examined. Note that the `WTVET` veteran population matches BLS by construction (it is the control total), so it is not an independent test.

**1. Setup: imports, file path, and the reference month in one place.**

````python
# Set up 
import  pandas as pd
df = pd.read_csv("../data/cps_00001.csv")

# Define the constants once
## Codes from the IPUMS CPS codebook
CIVILIAN = 1             # POPSTAT: adult civilian
IN_LABOR_FORCE = 2       # LABFORCE
EMPLOYED = [10, 12]      # EMPSTAT: at work; has job but not at work last week
UNEMPLOYED = [21, 22]    # EMPSTAT: unemployed
NONVETERAN, VETERAN = 1, 2   # VETSTAT

# Define function "labor_stats()" to estimate the population, labor force, employment, unemployment, and unemployment rate
def labor_stats(data, weight):
    pop = data[weight].sum()
    lf = data.loc[data['LABFORCE'] == IN_LABOR_FORCE, weight].sum()
    emp = data.loc[data['EMPSTAT'].isin(EMPLOYED), weight].sum()
    unemp = data.loc[data['EMPSTAT'].isin(UNEMPLOYED), weight].sum()
    return {"pop": pop, "lf": lf, "emp": emp, "unemp": unemp, "rate": unemp / lf * 100}
````

**3. National (headline) Estimates**

*Step 3a: Define the universe: civilian, age 16+*
Note: BLS's official labor force universe: civilian noninstitutional population, age 16+
For veteran status, BLS's official labor force universe: civilian noninstitutional population, age 18+


````python
# BLS's official labor force universe: civilian noninstitutional population, age 16+
civ_pop = df[(df['POPSTAT'] == CIVILIAN) & (df['AGE'] >= 16)]
civ_pop_size = civ_pop['WTFINL'].sum()

# In the labor force: LABFORCE == 2
labor_force = civ_pop[civ_pop['LABFORCE'] == IN_LABOR_FORCE]
labor_force_size = labor_force['WTFINL'].sum()

# Unemployed: EMPSTAT 21 or 22
unemployed = civ_pop[civ_pop['EMPSTAT'].isin(UNEMPLOYED)]
unemployed_count = unemployed['WTFINL'].sum()

unemployment_rate = unemployed_count / labor_force_size * 100

print(f"Civilian noninstitutional population 16+: {civ_pop_size:,.0f}")
print(f"Labor force size: {labor_force_size:,.0f}")
print(f"Unemployed: {unemployed_count:,.0f}")
print(f"Unemployment rate: {unemployment_rate:.1f}%")
````

````python
# Use the "labor_stats()" function defined in "SETUP" and loop over the two sample weights in question

for w in ['WTFINL', 'COMPWT']:
    r = labor_stats(civ_pop, w)
    print(f"{w}: pop={r['pop']:,.0f}  LF={r['lf']:,.0f}  "
          f"unemployed={r['unemp']:,.0f}  rate={r['rate']:.2f}%")

# Save the results produced by running the function
headline = {w: labor_stats(civ_pop, w) for w in ['WTFINL', 'COMPWT']}

# Print out the results
pd.DataFrame(headline).style.format("{:,.2f}")

# Display the same results in a table
display(pd.DataFrame(headline).style.format("{:,.1f}"))
````

````python
# Subset the results produced above to only the ones associated with the weight `COMPWT`
c = headline['COMPWT']

# Produce a comparison table: BLS published statistics vs. this analysis.
comparison = pd.DataFrame({
    "This analysis (COMPWT)": [c['pop'], c['lf'], c['unemp'], c['rate']],
    "BLS (NSA)": [267_884_000, 167_960_000, 6_604_000, 3.9],
}, index=["Population 16+", "Labor force", "Unemployed", "Unemployment rate (%)"])

# Display the comparison table
display(comparison.style.format("{:,.1f}"))
````

**4. Veteran and Nonveteran Esimates**

Now I want to replicate the BLS published numbers for the veteran and nonveteran sub-populations. The approach involves 1) completing the above exercise for subsets (i.e., by veteran status), and 2) confirming which sample weight is needed for the veteran and nonveteran populations. 

Note: It is difficult to confirm from BLS  documentation alone is whether BLS's own published veteran unemployment rate is computed using `WTVET`-reweighted data, or the standard weight (`WTFINL`/`COMPWT`) just filtered to veterans.


*4a. Inspect `VETSTAT`*

[IPUMS-CPS codecook] case count for `VETSTAT` (March 2024)
    0 (NIU): 18,544K | 1 (No service): 71,305 | 2 (Yes): 5,462

[`VETSTAT` variable values and description]
- 0 (NIU): NIU = Not In Universe. In CPS, this happens mainly because the respondent is under the minimum age used for veteran status questions.
- 1 (No service): the person was asked, and answered that they haven't served.
- 2 (Yes): the person was asked, and answered that they have served — this is your veterans group.
- 9 (Unknown): asked, but no usable answer (e.g., don't know/refused).

Not In Universe (NIU) means this question was never asked of this person, because they don’t fall into the population the question applies to. It’s the survey’s way of saying “not applicable,” which is different from “asked, and the answer was no.
- If you treated 0 as "not a veteran" and lumped it in with 1, you'd be including people the question was never even posed to; mostly people under 18. Since you're about to filter to AGE >= 18 anyway, this should mostly resolve itself, but it's worth confirming rather than assuming.
- Verification step: after filtering to your 18+ civilian population, see whether the 0 (NIU) count drops sharply, close to zero. That would confirm NIU is standing in for “too young to be asked,” and that your subgroup filter using codes 1 and 2 is clean once you’ve restricted to adults.

````python
#Step 4a - Inspect VETSTAT

counts = df['VETSTAT'].value_counts(dropna=False).sort_index()

codebook_labels = {0: "NIU", 1: "No service", 2: "Yes", 9: "Unknown"}

vetstat_table = pd.DataFrame({
    "Code": counts.index,
    "Codebook label": counts.index.map(codebook_labels),
    "This analysis": counts.values,
    "IPUMS (online)": [18_544, 71_305, 5_462],
}, index=[0,1,2])

vetstat_table["Match"] = vetstat_table["This analysis"] == vetstat_table["IPUMS (online)"]

display(
    vetstat_table.style
    .format({"This analysis": "{:,.0f}", "IPUMS (online)": "{:,.0f}"})
    .hide(axis="index")
)
````

*4a. Results from Inspect `VETSTAT`*
Verifcation successful. Yes, observations where [`VETSTAT` = 0 (NIU)] drops. This confirms that NIU is mostly people under 18 years old, meaning 1) they should *NOT* be lumped in with `VETSTAT` = 1 (No service); and 2) NIU observations drop out after filtering out observations that are < 18 years old.

````python
# Step 4b — Build the 18+ civilian universe
## For veteran status, BLS's official labor force universe: civilian noninstitutional population, age 18+

civ_pop_18 = df[(df['POPSTAT'] == CIVILIAN) & (df['AGE'] >= 18)]
civ_pop_18_size = civ_pop_18['WTFINL'].sum()

print(civ_pop_18['VETSTAT'].value_counts(dropna=False).sort_index())
````

````python
# Step 4c — Use the labor_stats() function again

## Create the two subpopulations by veteran status
veterans = civ_pop_18[civ_pop_18['VETSTAT'] == VETERAN]
nonveterans = civ_pop_18[civ_pop_18['VETSTAT'] ==  NONVETERAN]

## create the lists of subpops (groups) and sampling weights (weights) to use in the loops below
groups = {"Veterans": veterans, "Nonveterans": nonveterans}
weights =["WTFINL", "COMPWT", "WTVET"]

vresults = {}       # create an empty container to store results

# Use the "labor_stats()" function defined in "SETUP" and loop over the two sample weights in question
for a, b in groups.items():
    for wt in weights:
        vresults[f"{a} ({wt})"] = labor_stats(b, wt)     # save output from the function into a new column
        #print(f"population: {a} and weight: {wt}")      # show the column heading names

# Check
print(len(vresults))                 # should be 6
print(list(vresults))                # the six labels

# Does each label hold the right numbers? Recompute one combination from scratch.
key = "Veterans (WTFINL)"
print(vresults[key] == labor_stats(veterans, "WTFINL"))    # should be True


# Step 4d — Find the actual published comparison numbers. 
## Find the “Employment Situation of Veterans” release for your target month on bls.gov
### Veterans (NSA): pop = 17654; lf = 8556; emp = 8300; unemp = 257; ur = 3.0
### Veterans (SA):  pop = ?????; lf = 8530; emp = 8280; unemp = 249; ur = 2.9
### Nonveterans (NSA): pop = 241051; lf = 157149; emp = 151111; unemp = 6038; ur = 3.8
### Nonveterans (SA):  pop = ??????; lf = 156845; emp = 150947; unemp = 5898; ur = 3.8

# Add the BLS published numbers
vresults["Veterans (BLS)"] = {"pop": 17_654_000,  "lf": 8_556_000,   "emp": 8_300_000,   "unemp": 257_000,  "rate": 3.0}
vresults["Nonveterans (BLS)"] = {"pop": 241_051_000, "lf": 157_149_000, "emp": 151_111_000, "unemp": 6_038_000, "rate": 3.8}

# Turn the results into a table
vres = pd.DataFrame(vresults)    

# Reorder the columns in the table to the way I want it. In this case, it's moving the BLS columns from the last 2 cols.
## State the new order I want
column_order = [
    "Veterans (BLS)", "Veterans (WTFINL)", "Veterans (COMPWT)", "Veterans (WTVET)", 
    "Nonveterans (BLS)", "Nonveterans (WTFINL)", "Nonveterans (COMPWT)", "Nonveterans (WTVET)", 
]

## Assign that order of columns to the table
vres = vres[column_order]
display(vres.style.format("{:,.1f}")) # Show the table with format
````


**Why I changed it.** The loop version removes the six repeated lines, so adding a weight or a group means editing one list. It also puts the BLS columns in the same table as my results, so the percent-difference loop reads everything from one place.

**What I kept in mind.** The explicit version is easier to read line by line. The loop is more compact, but it relies on labels being built consistently (`f"{group} ({weight})"`).

**Lessons from getting it working.**
- To combine two existing tables, use `pd.concat([a, b], axis=1)`. `pd.DataFrame({a, b})` fails because `{a, b}` is a set, not a dictionary.
- BLS figures are in thousands, so convert before comparing.
- Percent difference = (this analysis - BLS) / BLS, so positive means my estimate is higher.

````python
# Step 4e — Compare estimates using weights `WTFINL', `COMPWT', and `WTVET' against benchmark (BLS published numbers)
    
pct_diff = {}
for group in ["Veterans", "Nonveterans"]:
    for weight in ["WTFINL", "COMPWT", "WTVET"]:
        mine = vres[f"{group} ({weight})"]
        bls = vres[f"{group} (BLS)"]
        pct_diff[(group, weight)] = (mine - bls) / bls * 100

# Turn pct_diff into a table
pct_table = pd.DataFrame(pct_diff)

## Drop the unemployment rate
pct_table = pct_table.drop(index="rate")

## Show table
display(pct_table.style.format("{:+.2f}%"))
````

**Results** `WTVET` is the closest match for both groups. The veteran population matches BLS *by construction* (it is the control total the weight is built to hit), so it is not an independent test; the labor force, employed, and unemployed rows are.
